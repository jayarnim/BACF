```python
MODEL_CLS = "bamf"
COMB = "cat"
DATASET = "abeauty"
STRATEGY = "pointwise"
PATH_WORK_DIR = "../../.."
```

```python
if DATASET=="abeauty":
    NUM_USERS = 3819
    NUM_ITEMS = 1581
    FILTER = "default"
    MAX_LEN = None
elif DATASET=="amusic":
    NUM_USERS = 5541
    NUM_ITEMS = 3568
    FILTER = "default"
    MAX_LEN = None
elif DATASET=="lastfm":
    NUM_USERS = 1892
    NUM_ITEMS = 17632
    FILTER = "default"
    MAX_LEN = None
elif DATASET=="movielens":
    NUM_USERS = 610
    NUM_ITEMS = 9724
    FILTER = "tfidf"
    MAX_LEN = 400
else:
    raise ValueError()

ANCHOR = "user"
```

# Config

```python
import pandas as pd
from components import histories
from pointwiseBayes import config, datamodule, trainer, evaluator, utils
import bacf
```

```python
PATH = f"./config/{MODEL_CLS}_{STRATEGY}.yaml"
cfg = config.build(PATH)
```

```python
utils.seed.main(cfg.seed)
```

# Data Preparation

```python
PATH = f"./data/{DATASET}.csv"
df = pd.read_csv(PATH)
```

```python
kwargs = dict(
    df=df,
    percentile=0.9,
)

utils.desc.main(**kwargs)
```

```python
kwargs = dict(
    df=df,
    cfg=cfg.datamodule,
)

dataloaders, trn = datamodule.build(**kwargs)
```

```python
kwargs = dict(
    df=trn,
    num_users=NUM_USERS,
    num_items=NUM_ITEMS,
    anchor=ANCHOR,
    filter=FILTER,
    max_len=MAX_LEN,
)

histories_cls = histories.Histories(**kwargs)
```

# Modeling

```python
kwargs = dict(
    histories=histories_cls,
    num_users=NUM_USERS,
    num_items=NUM_ITEMS,
    comb=COMB,
    **cfg.model.params,
)

model = bacf.bamf.BayesianAttentionalMatrixFactorization(**kwargs)
```

# Train

```python
kwargs = dict(
    model=model,
    cfg=cfg.trainer,
)

opt = trainer.build(**kwargs)
```

```python
kwargs = dict(
    trn_loader=dataloaders["trn"],
    val_loader=dataloaders["val"],
)

records = opt.fit(**kwargs)
```

```python
kwargs = dict(
    records=records,
    suptitle="BAMF Training History",
    trn="BCE",
    val="NDCG@10",
    figsize=(7,3),
)

utils.visualizer.main(**kwargs)
```

# Test

```python
kwargs = dict(
    model=model,
    cutoff=[5,10,15,20,25,50],
)

msr = evaluator.build(**kwargs)
```

```python
result, metrics_sheet = msr(dataloaders["tst"])
```
